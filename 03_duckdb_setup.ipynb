```python
import duckdb
import pandas as pd
```

```python
# Reload from CSV to ensure clean data
books_clean_table = pd.read_csv('books_clean.csv')
reviews_clean_table = pd.read_csv('reviews_clean.csv')
author_genders_table = pd.read_csv('author_genders.csv')
```

```python
conn = duckdb.connect('amazon_books.ddb')
```

```python
# Create database

conn.execute("CREATE TABLE books AS SELECT * FROM books_clean_table")
conn.execute("CREATE TABLE reviews AS SELECT * FROM reviews_clean_table")

conn.execute("CREATE TABLE author_genders AS SELECT * FROM author_genders_table")
```

```python
# Verify
for table in ['books', 'reviews', 'author_genders']:
    count = conn.execute(f"SELECT COUNT(*) FROM {table}").fetchone()[0]
    print(f"{table:20} → {count:,} rows")
```

```python
conn.close()
```
